```python
#read_csv(filepath)
import pandas as pd
import numpy as np

filepath ='https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv'
df = pd.read_csv(filepath)
df
```

```python
# Shape
print("Shape of dataset:")
print(df.shape)
print("\n")

# First 5 rows
print("\nFirst 5 rows:")
print(df.head())
print("\n")

# Column names
print("\nColumn names:")
print(df.columns)
print("\n")

# Data types
print("\nData types:")
print(df.dtypes)
print("\n")

# Missing values
print("\nMissing values:")
print(df.isnull().sum())
print("\n")

# 4) To Check and remove duplicates.

# Check number of duplicate rows
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
print("\n")

# Remove duplicates
df = df.drop_duplicates()

print("\nShape after removing duplicates:")
print(df.shape)
print("\n")

# 5) Rename column "Title" to "Movie_Title"

df.rename(columns={'Title': 'Movie_Title'}, inplace=True)

print("\nColumns after renaming:")
print(df.columns)
print("\n")

# 6) Get unique genres.

print("\nUnique Genres:")
print(df['Genre'].unique())
print("\n")

# 7) Extract number of genres per movie

print(df['Genre'].str.split(',').str.len())
print("\n")
print()

#OR

print(df[['Movie_Title', 'Genre']].assign(
    Number_of_Genres=df['Genre'].str.split(',').str.len()))
print()

# 8)  Handle missing values in 'Revenue (Millions)' and 'Metascore.

df['Revenue (Millions)'] = df['Revenue (Millions)'].fillna(df['Revenue (Millions)'].mean()) # fillna() - Use to replace/handle missing values.

df['Metascore'] = df['Metascore'].fillna(df['Metascore'].mean())

print(df.isnull().sum())
print()

# 9) Top 5 movies by Rating.

top5 = df.sort_values('Rating', ascending=False).head(5)

print(top5[['Movie_Title', 'Rating']])

print()

# 10) Movie with highest revenue

highest = df.sort_values('Revenue (Millions)', ascending=False).head(1)
print("Movie with highest and lowest revenue: ")

print(highest[['Movie_Title', 'Revenue (Millions)']])
print()

# Movie with lowest revenue.

lowest = df.sort_values('Revenue (Millions)').head(1)
print("Movie with highest and lowest revenue: ")

print(lowest[['Movie_Title', 'Revenue (Millions)']])
print()

#11) Average rating by Director.

avg_rating = df.groupby('Director')['Rating'].mean()
print("Average rating by Director: ")
print()
print(avg_rating)
print()

#12) Movies per year.

movies_year = df.groupby('Year')['Movie_Title'].count()
print("Movies per year: ")
print()
print(movies_year)
print()

#13)  Movies with rating > 8 & revenue > 100M

result = df[(df['Rating'] > 8) & (df['Revenue (Millions)'] > 100)]
print("Movies with rating > 8 & revenue > 100M: ")
print()
print(result[['Movie_Title', 'Rating', 'Revenue (Millions)']])
print()

#14) Group by Director: avg Rating, Revenue.

director_group = df.groupby('Director')[['Rating', 'Revenue (Millions)']].mean()
print("Group by Director: avg Rating, Revenue: ")
print()
print(director_group)
print()

#15)  Group by Year: count, avg Rating.

year_group = df.groupby('Year')[['Rating']].agg(['count', 'mean']) # Performs multiple calculation on 'Rating' column.
print("Group by Year: count, avg Rating: ")
print()
print(year_group)
print()

#16) Create a new column Profitability:
# a) Revenue divided by Duration.

df['Profitability'] = df['Revenue (Millions)'] / df['Runtime (Minutes)']
print("New column Profitability: ")
print()
print(df[['Movie_Title', 'Profitability']])
print()



top10 = df.sort_values('Profitability', ascending=False).head(10)
print("Top 10 most profitable movies: ")
print()
print(top10[['Movie_Title', 'Profitability']])
print()


```
